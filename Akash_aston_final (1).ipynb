#Feature Engineering

Ensure Timestamp Format

```python
df_model['timestamp'] = pd.to_datetime(df_model['timestamp'])
```

Sort Data

```python
df_model = df_model.sort_values(by='timestamp')
```

Create Time-Based Features

```python
df_model['hour'] = df_model['timestamp'].dt.hour
df_model['day'] = df_model['timestamp'].dt.day
df_model['month'] = df_model['timestamp'].dt.month
df_model['weekday'] = df_model['timestamp'].dt.weekday
```

Create Lag Features

```python
df_model['NO2_lag1'] = df_model['NO2'].shift(1)
df_model['NO2_lag2'] = df_model['NO2'].shift(2)
```

```python
print(df_model.columns.tolist())
print(df_model.shape)
display(df_model.head())
```

```python
# ============================================================
# VALIDATION EXPERIMENT
# Corrected historical NO2 features
# ============================================================

df_validation = df_model.copy()

# Convert timestamp to datetime
df_validation['timestamp'] = pd.to_datetime(
    df_validation['timestamp'],
    errors='coerce'
)

# Sort by monitoring station and time
df_validation = df_validation.sort_values(
    ['SiteName', 'timestamp']
).reset_index(drop=True)

# ------------------------------------------------------------
# Correct lag features
# ------------------------------------------------------------

df_validation['NO2_lag1_corrected'] = (
    df_validation.groupby('SiteName')['NO2']
    .shift(1)
)

df_validation['NO2_lag2_corrected'] = (
    df_validation.groupby('SiteName')['NO2']
    .shift(2)
)

# ------------------------------------------------------------
# Correct rolling mean
# Previous 3 observations ONLY
# Current NO2 is excluded
# ------------------------------------------------------------

df_validation['NO2_roll3_corrected'] = (
    df_validation
    .groupby('SiteName')['NO2']
    .shift(1)
    .groupby(df_validation['SiteName'])
    .rolling(window=3)
    .mean()
    .reset_index(level=0, drop=True)
)

print("Corrected features created successfully.")

display(
    df_validation[
        [
            'timestamp',
            'SiteName',
            'NO2',
            'NO2_lag1_corrected',
            'NO2_lag2_corrected',
            'NO2_roll3_corrected'
        ]
    ].head(10)
)
```

```python
original_corr = None

if 'NO2_roll3' in df_validation.columns:
    original_corr = df_validation['NO2'].corr(
        df_validation['NO2_roll3']
    )

corrected_corr = df_validation['NO2'].corr(
    df_validation['NO2_roll3_corrected']
)

print("Original NO2_roll3 correlation:", original_corr)
print("Corrected NO2_roll3 correlation:", corrected_corr)
```

```python
# ============================================================
# MODEL EVALUATION FUNCTION
# ============================================================

def run_experiment(
    X_train,
    X_test,
    y_train,
    y_test,
    experiment_name
):

    models = {

        'Linear Regression':
            LinearRegression(),

        'Decision Tree':
            DecisionTreeRegressor(
                random_state=42
            ),

        'Random Forest':
            RandomForestRegressor(
                n_estimators=100,
                random_state=42,
                n_jobs=-1
            ),

        'XGBoost':
            XGBRegressor(
                n_estimators=100,
                max_depth=6,
                learning_rate=0.1,
                random_state=42,
                n_jobs=-1,
                objective='reg:squarederror'
            )
    }

    results = []

    for model_name, model in models.items():

        print(
            f"Running {model_name} "
            f"for {experiment_name}..."
        )

        model.fit(
            X_train,
            y_train
        )

        predictions = model.predict(
            X_test
        )

        mae = mean_absolute_error(
            y_test,
            predictions
        )

        rmse = np.sqrt(
            mean_squared_error(
                y_test,
                predictions
            )
        )

        r2 = r2_score(
            y_test,
            predictions
        )

        results.append({
            'Experiment': experiment_name,
            'Model': model_name,
            'MAE': mae,
            'RMSE': rmse,
            'R2': r2
        })

    return pd.DataFrame(results)
```

```python
# ============================================================
# MODEL EVALUATION FUNCTION
# ============================================================

def run_experiment(
    X_train,
    X_test,
    y_train,
    y_test,
    experiment_name
):

    models = {

        'Linear Regression':
            LinearRegression(),

        'Decision Tree':
            DecisionTreeRegressor(
                random_state=42
            ),

        'Random Forest':
            RandomForestRegressor(
                n_estimators=30,
                max_depth=10,
                n_jobs=-1,
                random_state=42
            ),

        'XGBoost':
            XGBRegressor(
                n_estimators=100,
                learning_rate=0.1,
                random_state=42
            )
    }

    results = []

    for model_name, model in models.items():

        print(
            f"Running {model_name} - "
            f"{experiment_name}"
        )

        model.fit(
            X_train,
            y_train
        )

        predictions = model.predict(
            X_test
        )

        mae = mean_absolute_error(
            y_test,
            predictions
        )

        rmse = np.sqrt(
            mean_squared_error(
                y_test,
                predictions
            )
        )

        r2 = r2_score(
            y_test,
            predictions
        )

        results.append({
            'Experiment': experiment_name,
            'Model': model_name,
            'MAE': mae,
            'RMSE': rmse,
            'R2': r2
        })

    return pd.DataFrame(results)
```

#Spatial Clustering (Hotspot Detection)

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import DBSCAN
```

Standardising Latitude and Longitude

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

coords_scaled = scaler.fit_transform(
    df_model[['latitude', 'longitude']]
)
```

Apply K-Means Clustering

```python
from sklearn.cluster import KMeans

# Apply K-Means clustering
kmeans = KMeans(
    n_clusters=5,
    random_state=42,
    n_init=10
)

clusters = kmeans.fit_predict(coords_scaled)

# Add cluster labels to the dataset
df_model['Cluster'] = clusters

# Display the first few records
df_model[['SiteName', 'latitude', 'longitude', 'Cluster']].head()
```

Calculate Average NO₂ Concentration for Each Cluster

```python
# Calculate average NO2 concentration for each cluster
cluster_summary = (
    df_model.groupby('Cluster')['NO2']
    .mean()
    .reset_index()
    .sort_values('NO2', ascending=False)
)

print(cluster_summary)
```

Identify the Hotspot Cluster

```python
# Identify the hotspot cluster
hotspot_cluster = cluster_summary.iloc[0]['Cluster']

print(f"Hotspot Cluster: {hotspot_cluster}")
```

```python
import matplotlib.pyplot as plt

# -------------------------------------------------
# K-Means Hotspot Visualisation
# -------------------------------------------------

# Make a copy of the clustered data
plot_data = df_model.copy()

# The K-Means cluster labels should already exist
# If your column has a different name, change 'Cluster'
cluster_column = 'Cluster'

# Calculate mean NO2 for each cluster
cluster_mean_no2 = (
    plot_data.groupby(cluster_column)['NO2']
    .mean()
    .sort_values(ascending=False)
)

# Identify the cluster with the highest mean NO2
hotspot_cluster = cluster_mean_no2.idxmax()

print("Mean NO2 concentration by cluster:")
print(cluster_mean_no2)

print("\nHotspot Cluster:", hotspot_cluster)
print("Mean NO2:", round(cluster_mean_no2.loc[hotspot_cluster], 2), "µg/m³")


# -------------------------------------------------
# Plot spatial clusters
# -------------------------------------------------

plt.figure(figsize=(12, 8))

# Plot non-hotspot clusters
for cluster in sorted(plot_data[cluster_column].unique()):

    cluster_data = plot_data[
        plot_data[cluster_column] == cluster
    ]

    if cluster == hotspot_cluster:
        continue

    plt.scatter(
        cluster_data['longitude'],
        cluster_data['latitude'],
        s=8,
        alpha=0.5,
        label=f'Cluster {cluster}'
    )


# Plot hotspot cluster separately
hotspot_data = plot_data[
    plot_data[cluster_column] == hotspot_cluster
]

plt.scatter(
    hotspot_data['longitude'],
    hotspot_data['latitude'],
    s=12,
    alpha=0.7,
    label=f'Hotspot - Cluster {hotspot_cluster}'
)

plt.xlabel("Longitude")
plt.ylabel("Latitude")

plt.title(
    f"K-Means NO₂ Pollution Hotspot Detection "
    f"(Cluster {hotspot_cluster})"
)

plt.legend()
plt.grid(True, alpha=0.2)

plt.tight_layout()

# Save the image
plt.savefig(
    "hotspot_map.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
import matplotlib.pyplot as plt
import os

# ============================================================
# FINAL CORRECTED MODEL RESULTS
# ============================================================

model_results = {
    "Linear Regression": {
        "MAE": 4.0434,
        "RMSE": 6.1986,
        "R²": 0.8405
    },

    "Decision Tree": {
        "MAE": 6.1973,
        "RMSE": 9.3013,
        "R²": 0.6409
    },

    "Random Forest": {
        "MAE": 3.8748,
        "RMSE": 5.9533,
        "R²": 0.8529
    },

    "XGBoost": {
        "MAE": 3.9731,
        "RMSE": 6.0510,
        "R²": 0.8480
    }
}

# Folder for saving figures
os.makedirs("final_model_figures", exist_ok=True)


# ============================================================
# FUNCTION TO CREATE INDIVIDUAL MODEL FIGURE
# ============================================================

def create_model_figure(model_name, metrics):

    metric_names = ["MAE", "RMSE", "R²"]
    values = [
        metrics["MAE"],
        metrics["RMSE"],
        metrics["R²"]
    ]

    plt.figure(figsize=(8, 5))

    bars = plt.bar(metric_names, values)

    plt.title(
        f"{model_name} Regression Performance",
        fontsize=14,
        fontweight="bold"
    )

    plt.xlabel("Evaluation Metric", fontsize=11)
    plt.ylabel("Value", fontsize=11)

    # Add values above bars
    for bar, value in zip(bars, values):

        plt.text(
            bar.get_x() + bar.get_width() / 2,
            bar.get_height() + 0.15,
            f"{value:.4f}",
            ha="center",
            va="bottom",
            fontsize=10
        )

    plt.tight_layout()

    # Safe filename
    filename = model_name.lower().replace(" ", "_").replace("²", "2")

    filepath = os.path.join(
        "final_model_figures",
        f"{filename}_regression_performance.png"
    )

    plt.savefig(
        filepath,
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()

    print(f"Saved: {filepath}")


# ============================================================
# GENERATE FIGURES FOR ALL MODELS
# ============================================================

for model_name, metrics in model_results.items():

    create_model_figure(
        model_name,
        metrics
    )
```
